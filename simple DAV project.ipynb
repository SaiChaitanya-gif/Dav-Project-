# Data cleaning

Create a backup

```python
weather = df.copy()
```

Remove duplicate records

```python
weather.drop_duplicates(inplace=True)
```

Check missing values

```python
weather.isnull().sum()
```

Handle missing numerical values

```python
num_cols = weather.select_dtypes(
    include=np.number
).columns

for col in num_cols:
    weather[col] = weather[col].fillna(
        weather[col].median()
    )
    #Median is less affected by extreme values than the mean
```

Handle missing categorical values

```python
cat_cols = weather.select_dtypes(
    include='object'
).columns

for col in cat_cols:
    if weather[col].isnull().sum() > 0:
        weather[col] = weather[col].fillna(
            weather[col].mode()[0]
        )
```

Remove extra spaces from column names

```python
weather.columns = weather.columns.str.strip()
```

Check cleaning results

```python
print("Missing values:")
print(weather.isnull().sum())

print("Duplicate values:")
print(weather.duplicated().sum())
```

# Data visualization using Matplotlib and Seaborn

Temperature distribution

```python
plt.figure(figsize=(10,5))

sns.histplot(
    data=weather,
    x='temperature_c',
    bins=30,
    kde=True
)

plt.title('Temperature Distribution')
plt.xlabel('Temperature (°C)')
plt.ylabel('Frequency')

plt.show()
```

Temperature trend over time

```python
plt.figure(figsize=(14,5))

plt.plot(
    weather['date_time'],
    weather['temperature_c']
)

plt.title('Temperature Trend Over Time')
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

Humidity distribution

```python
plt.figure(figsize=(10,5))

sns.histplot(
    data=weather,
    x='relative_humidity_pct',
    bins=30,
    kde=True
)

plt.title('Relative Humidity Distribution')
plt.xlabel('Relative Humidity (%)')

plt.show()
```

Temperature vs Humidity

```python
plt.figure(figsize=(10,5))

sns.scatterplot(
    data=weather,
    x='temperature_c',
    y='relative_humidity_pct',
    alpha=0.5
)

plt.title('Temperature vs Relative Humidity')
plt.xlabel('Temperature (°C)')
plt.ylabel('Humidity (%)')

plt.show()
```

Weather condition frequency

```python
plt.figure(figsize=(12,6))

sns.countplot(
    data=weather,
    y='weather',
    order=weather['weather'].value_counts().index
)

plt.title('Weather Condition Frequency')
plt.xlabel('Count')
plt.ylabel('Weather Condition')

plt.tight_layout()
plt.show()
```

Correlation heatmap

```python
plt.figure(figsize=(12,8))

numeric_data = weather.select_dtypes(
    include=np.number
)

correlation = numeric_data.corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Weather Data Correlation Heatmap')

plt.show()
```
